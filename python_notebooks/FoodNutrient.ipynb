## 6. Algorithm Selection Motivation

Since we observed an overlapped and pretty mixed cluster structure in the TSNE plot, we think a Soft Assignment Algorithms such as Fuzzy C-Means and Gaussian Mixture Models may work better here.

We see that there may be some overlap between some clusters in our dataset, so fuzzy c-means can be useful as it will help reflect the overlapping nature of these clusters. Evaluating the fuzzy clustering membership scores will help us answer whether there are some food categories that are more easily confused with others.


For GMM, it is a relatively flexible clustering algorithm that allows for overlapping clusters and diverse shapes, and does not require prior knowledge of the fuzziness of clusters, it also can generate new observations based on current cluster structure, leaving a potential door for new food product development.

## 7. Fuzzy C-Means

### 7.1. Parameter Selection

- We need to determine c the number of clusters we want and the fuzziness parameter p (represented as m in Python).

- The goal of this analysis is to identify the 14 food category and to understand more about the overlapping nature of how food categories are represented based on nutrition fact. Thus, we would choose c=14 and different p at random state 88 with the help of our t-SNE plot to see which parameter values will best help us meet this goal.

```python
c=14

for p in np.linspace(1.01, 2, 6):
    print('------------------------------------------')
    print('Analysis for c=%s and p=%s'%(c,p))
    centroids, W, W_0, dist_mat, history, iters, ndpc = cmeans(X.T, c=c, m=p, error=1e-6, maxiter=20000, seed=88)

    #Color code t-SNE Plots
    fig, ax = plt.subplots(2, 7, figsize=(20, 10))
    plt.suptitle('Fuzzy Clustering: %s clusters and p=%s'%(c,p), fontsize=14)
    for i in range(0,14):

        #Create a column with cluster i membership scores
        df_combo['Cluster %s Membership Score'%str(i)]=W[i,:]

        #Plot the points, color coded by the membership score.
        points = plt.scatter(df_combo["x_projected"], df_combo["y_projected"],
                     c=df_combo['Cluster %s Membership Score'%str(i)],
                             s=30, cmap="BuGn",
                            vmin=0,
                            vmax=1)
        ax[int(np.floor(i/7)),int(i%7)].scatter(df_combo["x_projected"], df_combo["y_projected"],
                     c=df_combo['Cluster %s Membership Score'%str(i)],
                                                s=30, cmap="BuGn",
                                               vmin=0,
                                               vmax=1)

        fig.colorbar(points, ax=ax[int(np.floor(i/7)),int(i%7)])
        ax[int(np.floor(i/7)),int(i%7)].set_title('Cluster %s Membership Scores'%(str(i)))
    plt.show()
```

The fuzzy clustering produced by Fuzzy c-Means with c=14 and p=1.208 best meets the goals. It is able to identify many of the clusters and subclusters suggested by the t-SNE plots. Furthermore, objects and clusters that are more well-separated from other clusters suggested by the t-SNE plots end up having high cluster membership weights to just that one cluster (suggested by the t-SNE plot). Furthermore, objects and clusters that are closer or less-separated from other clusters suggested by the t-SNE plots end up having higher cluster membership weights to multiple clusters, reflecting the uncertainty of these assignments.

### 7.4. Assessing Clustering Separation and Cohesion

Create a hard partition by assigning each object to the cluster that it most belongs to.

```python
partition_cluster_labels=[df_membership.T[i].argmax() for i in range(0,len(df_membership))]
partition_cluster_labels[:20]
```

```python
df_combo['predicted_cluster']=partition_cluster_labels
df_combo.head()
```

```python
def show_silhouette_plots(X,cluster_labels):

    
    #How many clusters in your clustering?
    n_clusters=len(np.unique(cluster_labels))    

    # Create a subplot with 1 row and 2 columns
    fig, ax1 = plt.subplots(1, 1)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot
    # The silhouette coefficient fcan range from -1, 1 but in this example all
    # lie within [-0.1, 1]
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(X) + (n_clusters + 1) * 10])


    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(X, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(X, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]
    

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot
        y_lower = y_upper + 10  # 10 for the 0 samples

        ax1.set_title("The silhouette plot for the various clusters.")
        ax1.set_xlabel("The silhouette coefficient values")
        ax1.set_ylabel("Cluster label")

        # The vertical line for average silhouette score of all the values
        ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

        ax1.set_yticks([])  # Clear the yaxis labels / ticks
        ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])


    plt.show()
    
    return
```

```python
show_silhouette_plots(X, df_combo['predicted_cluster'])
```

- We can see that clusters 0, 1, 4, 6, 7, 8, 9, 11, 12, 13) have poor cohesion based on the more of a dramatic decline in their silhouette plot. They also contain objects (the ones with negative silhouette score values) that are closer to other clusters than the ones that they were assigned to. 
- Clusters 6 is the largest cluster in size, and the best regarding cohesion and separation.
- The average silhouette score being 0.27 means that the overall separation and cohesion is low.

AIC, BIC

```python
n = X2.shape[0]
d = X2.shape[1]
k = n / (41*(1+d+d^2))
k
```

```python
aics = []
bics = []

for k in range(1, 10+1):
    gm = GaussianMixture(n_components=k, random_state=88)
    gm.fit(X2)

    aics.append(gm.aic(X2))
    bics.append(gm.bic(X2))

print(aics)
print('------------------')
print(bics)
```

```python
fig, ax = plt.subplots(1,2, figsize=(20,7))

#AIC
ax[0].plot(range(1, 10+1), aics)
ax[0].set_xticks(range(1, 10+1))
ax[0].set_xlabel('k', fontsize=15)
ax[0].set_title('AIC', fontsize=16)

#BIC
ax[1].plot(range(1, 10+1), bics)
ax[1].set_xticks(range(1, 10+1))
ax[1].set_xlabel('k', fontsize=15)
ax[1].set_title('BIC', fontsize=16)

plt.show()
```

It seems that AIC and BIC does not really work here. What about the Silhouette Plot with each GMM model, k from 2-10?

Silhouette Plot

```python
K_list = range(2, 10+1)
trial_num = 3

avg_avg_sil_score_list = []
for k in K_list:
    avg_sil_score_list = []

    for i in range(0, trial_num):
        gm = GaussianMixture(n_components=k, random_state=88)
        gm.fit(X2)
        W = gm.predict_proba(X2)
        cluster_labels = np.argmax(W, axis=1).tolist()     #convert the GMM result to a hard-assignment
        avg_sil_score=silhouette_score(X2, cluster_labels)
        
        avg_sil_score_list.append(avg_sil_score)
    
    avg_avg_sil_score_list.append(np.average(avg_sil_score_list))
     
avg_avg_sil_score_list
```

```python
plt.figure(figsize=(7, 5))
plt.plot(K_list, avg_avg_sil_score_list)
plt.xlabel('Number of Clusters in GMM', fontsize=13)
plt.ylabel('Average of 3 Average Silhouette Score with K', fontsize=13)
plt.title('Average Silhouette Score Plot', fontsize=13)
plt.show()
```

```python
def show_silhouette_plots(X,cluster_labels):
    #How many clusters in your clustering?
    n_clusters=len(np.unique(cluster_labels))    

    # Create a subplot with 1 row and 2 columns
    fig, ax1 = plt.subplots(1, 1)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot
    # The silhouette coefficient fcan range from -1, 1 but in this example all
    # lie within [-0.1, 1]
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(X) + (n_clusters + 1) * 10])


    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(X, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(X, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]
    

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot
        y_lower = y_upper + 10  # 10 for the 0 samples

        ax1.set_title("The silhouette plot for the various clusters.")
        ax1.set_xlabel("The silhouette coefficient values")
        ax1.set_ylabel("Cluster label")

        # The vertical line for average silhouette score of all the values
        ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

        ax1.set_yticks([])  # Clear the yaxis labels / ticks
        ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])


    plt.show()
    
    return
```
